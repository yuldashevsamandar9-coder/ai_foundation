```python
import numpy as np

arr = np.array([1, 2, 3, 4])
print("nupmy_array:", arr),
arr.shape
```

```python
arr2 = np.array([[1, 2, 3, 4], [5, 6, 7, 9]])
print("nupmy_array:", arr2),
arr2.shape
```

```python
#dtype() datalar ustiida ishlashimizga yordam beradi
a= np.array([1, 2, 5])
print("nupmy_array:", a.dtype)
```

```python
#reshape shapeni uzgartirish uchun data turini
a = np.array([1,2,3,4,5,6])
b = a.reshape(2, 3)
print(b)
```

```python
#astype() data typeni uzgartirib beradi
a = np.array([2,3, 5])
b = a.astype(float)
print(b)
```

```python
#np.zeros np.ones() test datalarni yaratishda ishlatiladi => dummy.data
a = np.zeros((2,3))
b = np.ones((2,3))
print("zeros:", a),
print("ones:", b)
```

```python
#ndim data setimizda nechta ulchov borligini aniqlab beradi
a = np.array([[2,3,4], [9,6,8]])
print("ndim:", a.ndim)
```

```python
#arange pytondagi rangega uxshaydi malumotlar kam bulsa set data yaratishda
b = np.arange(1,7)
print("arange:", b)
```

```python
# sum(). #meann()  #max() #min()
a = np.array([1,0,2,3])
print("sum:", a.sum())
print("main:", a.mean())
print("max:", a.max())
print("min:", a.min())

```

```python
#flatten() 2D 3D arrayni 1d arrayfga almashtirib beradi
a = np.array([[1,2],[3,4]])
print("flatten:", a.flatten())
```

```python
# T transpose array malumotlarini qayta va urnini almashtirishda ham  tashkillashda ishlatiladi
a = np.array([[3,4],[5,6]])
print("T:", a.T)
```

```python
#dot() @ matritsalarni kupaytirishda neoron va networkda ham ishlatiladi
a = np.array([[1,2], [3,4]])
b = np.array([[2,0], [1,2]])
print("dot:", a.dot(b))

```

```python
#random.rand() random qilib yaratish
a = np.random.rand(2,3)
print("random.rand:", a)

```
