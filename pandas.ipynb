```python
data_with_missing = {
    "cars:": ["BWM", "TESLA", None, "VOLVO"],
    "counst": [8, 9, None, 4]
}
df.missing = pd.DataFrame(data_with_missing)
print(df.missing)
```

```python
df.missing.isnull().sum() #datalrni tozalashda ishlatiladi sum data bor yuglini aniqlaydi
```

```python
df.missing.dropna() # mauloti yugni uchirvoradi
```

```python
df.missing.dropna(how="all") # mauloti bittasi bulsa uchirmaydi
```

```python
df.missing.dropna(axis=1) #bitta ham malumot bulmasa hammasini uchirvoradi columnda turganlarni
```

```python
df.missing.dropna(subset=['counst']) # countsni tekshirib malumot bor yuglini aniqlaydi
```

```python
df.missing.fillna(0) # malumoti yug qatorga malumot quyib quyadi
```

```python
df.missing.fillna('unkwon') # malumoti yug qatorga malumot quyib quyadi
```

```python
df.missing.fillna({'cars:': 'unkwon', 'counst': 0}) # malumoti yug qatorga malumot quyib quyadi
```
